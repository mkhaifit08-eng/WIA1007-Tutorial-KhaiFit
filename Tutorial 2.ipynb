```python
import numpy as np #library for dealing with arrays/numbers
```

1. Create a numpy array of size 2 x 2 (2 rows and 2 columns)




```python
arr1 = np.array([[1, 2], [3, 4]])
print(arr1)
```

2. Create a random numpy array of size 3 x 4


```python
arr2 = np.random.rand(3,4)
print(arr2)
```

3. Create a numpy array of ones and a numpy array of zeroes


```python
arr3 = np.ones((3,3))
arr4 = np.zeros((3,3))
print("ones: ", arr3, "\n zeros: ", arr4)
```

4. Create a random numpy array of size 6x3 and transpose it


```python
arr5 = np.random.rand(6,3)
print(arr5)
arr6 = np.transpose(arr5)
print(arr6)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product


```python
arr7 = np.random.rand(3,2)
arr8 = np.random.rand(2,4)
print("array 1=\n", arr7)
print("array 1=\n", arr8)
arr9 = np.dot(arr7, arr8)
print("array 1 * array 2=\n", arr9)
```
