## Question 1

Question 1: Identify the Problem

Consider this dataset:

|Student |	Math | Science | English |
| ------ | ----- | ------- | ------- |
| Ali	| 75 | 80 | 72
| Sara | 88 | 91 | 85
| John | 65 | 70 | 68


1. Is this data in tidy format?
2. If not, explain which tidy-data rule is violated.
3. Using Python/pandas, reshape the dataset so that it has the columns:
    - Student
    - Subject
    - Score

### Solution

#### 1. Is this data in tidy format?

Answer: No, because it violates one of the tidy data principles 

#### 2. If not, explain which tidy-data rule is violated.

Answer: Variables are in columns.
Math, Science and English represent values of the same variable, Subject, but are stored in separate columns.

#### 3. Using Python/pandas, reshape the dataset so that it has the columns:
- Student
- Subject
- Score

Answer: Refer to code block below

```python
Q1data = pd.DataFrame({
    "Student": ["Ali", "Sara", "John"],
    "Math": [75,88,65],
    "Science": [80,91,70],
    "English": [72,85,68],
})

print(" Original Data: ")
print(Q1data)

Q1data_tidied = pd.melt(
    Q1data,
    id_vars = ["Student"],
    var_name = "Subject",
    value_name = "Score")

print("\nTidied Data: ")
print(Q1data_tidied)
```

## Question 2

Question 2: Sales Data

You are given:
|Country  | Q1_Sales | Q2_Sales |  Q3_Sales | Q4_Sales|
|  -----  | :------: | :------: | :-------: | :-----: |
|Malaysia |  12000   |  15000   |   14000   |  18000  |
|Thailand |  10000   |  11000   |   13000   |  16000  |
|Indonesia|  14000   |   12500  |   15000   |  17000  |


1. What is the tidy-data problem with this dataset?
2. Use pandas to convert it into:

| Country  | Quarter | Sales |
| -------- | :-----: | :----: |
| Malaysia |  Q1     | 12000 |
| Malaysia |  Q2     | 15000 |

### Solution

#### 1. What is the tidy-data problem with this dataset?

Answer: Q1, Q2, Q3 and Q4 represent values of the same variable, Quarter, but are stored in separate columns. The rule "Variables are in columns" is violated.

#### 2. Use pandas to convert it into:

| Country  | Quarter | Sales |
| -------- | :-----: | :----: |
| Malaysia |  Q1     | 12000 |
| Malaysia |  Q2     | 15000 |

```python
Q2data = pd.DataFrame({
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000,10000,14000],
    "Q2_Sales": [15000,11000,12500],
    "Q3_Sales": [14000,13000,15000],
    "Q4_Sales": [18000,16000,17000],
})

print(" Original Data: ")
print(Q1data)

Q2data_tidied = pd.melt(
    Q2data,
    id_vars = ["Country"],
    var_name = "Quarter",
    value_name = "Sales")

print("\nTidied Data: ")
print(Q2data_tidied)
```

## Question 3

Question 3: Asking Questions from Data
Consider this dataset:
data = {    "Student": ["A", "B", "C", "D", "E", "F"],    "Study_Hours": [2, 5, 3, 7, 4, 8],    "Score": [55, 72, 60, 85, 68, 90]}


Using Python:
1. What is the average study time?
2. What is the average score?
3. Which student obtained the highest score?
4. Which student studied the most?
5. Question: Does studying more hours appear to be associated with a higher score?
6. Create a scatter plot of Study_Hours vs Score.

```python
# Creating a dataset using python 
data = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
    })

print(data, "\n\n")

print(" 1. What is the average study time?")
average_Study_Hours = data["Study_Hours"].mean()
print(f"The average study hours is {average_Study_Hours} hours")
print()

print("2. What is the average score?")
average_score = data["Score"].mean()
print(f"The average score is {average_score} marks")
print()

print("3. Which student obtained the highest score?")
print(data[data["Score"] == data["Score"].max()])
student_highest_score = data.loc[data["Score"].idxmax(), "Student"]
print(f"Student {student_highest_score} obtained the highest score")
print()

print("4. Which student studied the most?")
print(data[data["Study_Hours"] == data["Study_Hours"].max()])
studied_most = data.loc[data["Study_Hours"].idxmax(), "Student"]
print(f"Student {studied_most} studied the most")
print()

print("6. Create a scatter plot of Study_Hours vs Score.")
data.plot.scatter(
    x="Study_Hours",
    y="Score",
    title="Scatter plot of Study Hours vs Score",
    grid=True,
)
```

5. Question: Does studying more hours appear to be associated with a higher score?

Answer: Yes, from the scatter plot and the pattern of data, studying more hours appear to be associated with a higher score
