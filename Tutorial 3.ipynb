### Choosing a Visualization

A line chart is suitable for showing how values change over time.

For this dataset:
- X-axis: date.
- Y-axis: daily total ridership.
- Each point: the total ridership on a particular date.
- Connecting lines: help show changes and patterns from day to day.

A line chart is appropriate because our research question concerns ridership changes throughout August 2026.


```python

# Select transport columns by position.
# Index 0 is the date column.
# The slice 1:15 selects indices 1 through 14.

daily_total = transport_data.iloc[:, 1:15].sum(axis=1)

print(daily_total)

# Create a line chart
plt.plot(transport_data['date'], daily_total)
plt.xlabel('Date')
plt.ylabel('Total Ridership')
plt.title('Daily Public Transport Ridership — August 2026')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Answer: The graph visualizes daily public transport ridership in August 2026. It shows how the total ridership changes from day to day.
